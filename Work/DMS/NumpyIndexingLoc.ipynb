```python
name=['Sneh','Bhavya','','Parth','Rehan','Uzair']
pin=[104,102,106,109,'',110]
city=['Ahmedabad','Surat','Pune','Nashik','Pune','']
w0=pd.DataFrame({'Name':name,'PIN':pin,'City':city})
print(w0,w0.isnull(),sep='\n')
```

```python
name1=['Sneh','Bhavya',np.nan,'Parth','Rehan','Uzair']
pin1=[104,102,106,109,np.nan,110]
city1=['Ahmedabad','Surat','Pune','Nashik','Pune',np.nan]
w1=pd.DataFrame({'Name':name1,'PIN':pin1,'City':city1})
print(w1,w1.isnull(),sep='\n')
```
