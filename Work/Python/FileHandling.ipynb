```python
# If "data.txt" does NOT exist: It creates the file and writes to it.
# If "data.txt" DOES exist: It crashes safely with a FileExistsError.
try:
    with open("data.txt", "x") as file:
        file.write("Hello, World!")
    print("File created successfully.")
except FileExistsError:
    print("Error: The file already exists! Data was protected.")
```

```python
def validate_pnr(pt):
    p=r'^\D{7}$'
    if re.fullmatch(p,pt):
        print('Valid',pt)
    else: print('Invalid',pt)
validate_pnr('123')
validate_pnr('1234567')
validate_pnr('abcdefg')
validate_pnr('abc')
```

 ##### Character Classes :

 [0-9] — any digit from 0 to 9
 
 [A-Z] — any uppercase letter
 
 [a-z] — any lowercase letter
 
 [A-Za-z] — any letter upper or lower
 
 [A-Za-z0-9] — any letter or digit

 
 ##### Special Sequences :

 \d — any digit (same as [0-9])
 
 \w — any word character (letters, digits, underscore)
 
 \s — any whitespace character
 
 \D — any non-digit character

 ##### Quantifiers :

 {n} — exactly n times
 
 {n,m} — between n and m times
 
 '+' — one or more times
 
 '*' — zero or more times
 
 '?' — zero or one time

 '.' — any charachter at that position --- if used with{n} then that character should repeat that times

```python
import re

def extract_train_numbers(text):
    pattern = r'\b[0-9]{4,5}\b'
    results = re.findall(pattern, text)
    print("Train Numbers Found :", results)
    return results
text = """
Booking confirmed for train 12951 Mumbai Rajdhani Express.
Passenger on train 12301 from Howrah to Delhi.
Train 12657 Bangalore Mail is on waitlist.
Coach number 4 berth 21 is assigned.
Train 19019 Dehradun Express departs at 06:20.
"""

extract_train_numbers(text)
```

```python
import re

class TicketValidator:

    def __init__(self, pnr, train_no, seat, fare):
        self.pnr      = pnr
        self.train_no = train_no
        self.seat     = seat
        self.fare     = fare

    def validate(self):
        print("=" * 40)
        print("   IndRail Ticket Validation Report")
        print("=" * 40)

        # Validate PNR - exactly 8 digits
        if re.fullmatch(r'[0-9]{8}', self.pnr):
            print("PNR      :", self.pnr, "-> VALID")
        else:
            print("PNR      :", self.pnr, "-> INVALID")

        # Validate Train Number - 4 or 5 digits
        if re.fullmatch(r'[0-9]{4,5}', self.train_no):
            print("Train No :", self.train_no, "-> VALID")
        else:
            print("Train No :", self.train_no, "-> INVALID")

ticket1 = TicketValidator("43821756", "12951", "B4",  "880")
ticket2 = TicketValidator("6793452",  "1230X", "b12", "XX")

ticket1.validate()
ticket2.validate()
```

```python
try:
    fare_amount = int(ticket1.fare)
    print("Ticket 1 Fare Conversion : SUCCESS")
except ValueError:
    print("Ticket 1 Fare Conversion : FAILED")

try:
    fare_amount = int(ticket2.fare)
    print("Ticket 2 Fare Conversion : SUCCESS")
except ValueError:
    print("Ticket 2 Fare Conversion : FAILED")
```
