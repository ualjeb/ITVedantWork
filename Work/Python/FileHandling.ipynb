```python
# If "data.txt" does NOT exist: It creates the file and writes to it.
# If "data.txt" DOES exist: It crashes safely with a FileExistsError.
try:
    with open("data.txt", "x") as file:
        file.write("Hello, World!")
    print("File created successfully.")
except FileExistsError:
    print("Error: The file already exists! Data was protected.")
```

```python
d=open('data.txt','w')
d.write('Hi')   #overwrite data by vanishing/deleting previous data
d.close()
e=open('data.txt','r')
f=e.read()
print(e,f)
f=e.read()  #pointer has reached last so empty string
print(e,f)
e.close()
print(e,f)
```

```python
with open('data.txt','a') as a: #this syntax in Python is called a Context Manager
    a.write('\nby')
#Its main purpose is to automatically close the file 
#as soon as the code inside the block finishes running, even if your code crashes or encounters an error midway.
with open('data.txt','r') as file:
    f=file.read()
print(f)
```
