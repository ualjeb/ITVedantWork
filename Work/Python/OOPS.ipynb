### Encapsulation

```python
class Car2:
    def __init__(self,amount):
        self.__amount=amount
    def total(self):
        print('Total Amount:',self.__amount)
    def depo(self,amt):
        self.__amount+=amt
        print('Finalize depo:',self.__amount)
t=Car2(10000)
t.total()
t.depo(200)
```
