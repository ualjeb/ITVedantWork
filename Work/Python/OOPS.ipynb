### Encapsulation

```python
class Car2:
    def __init__(self,amount):
        self.__amount=amount
    def total(self):
        print('Total Amount:',self.__amount)
    def depo(self,amt):
        self.__amount+=amt
        print('Finalize depo:',self.__amount)
t=Car2(10000)
t.total()
t.depo(200)
```

```python
print(t.__amount) #Private variable will not be accessible like this
```

### Inheritance
Inheritance is a core concept in Object-Oriented Programming (OOP) that allows a Child class (Subclass or DerivedClass) to reuse, extend and modify the attributes and methods of a Parent class (Base class).


Single: Child(Parent) → One parent class.

Multiple: Child(Parent1, Parent2) → More than one parent class.

Multilevel: Grandchild(Child) → Parent class itself inherits from another class.

Hierarchical: Child1(Parent), Child2(Parent) → Multiple children share one parent.

Hybrid: mix of all four type inheritance

#### Single Inheritance

```python
class Car:
    def move(self):
        print('The car is moving')
class BMW(Car):
    print('Always Done here')    # Python reads the script from top to bottom. When it encounters class BMW(Car):, it immediately executes the body of that class block to set it up.
    def run(self):
        print('Car is also here...')
r=Car()
r.move()
#r.run() will give error
```

```python
e=BMW()
e.run()
e.move()
```

```python
class Vehicle:
    def __init__(self,brand,model):
        self.brand=brand
        self.model=model
    def start_engine(self):
        return f'The engine of the {self.brand} {self.model} is now running'
class Car(Vehicle):
    def __init__(self,brand,model,number_of_doors):
        super().__init__(brand,model)# Use super() to call parent constructor
        self.number_of_doors=number_of_doors
    def open_trunk(self):
        return f'Opening the trunk of this {self.number_of_doors}-door car.'
my_car=Car('Toyota','Camry',4)
print(my_car.start_engine())#Inherited method
print(my_car.open_trunk())#Child-specific method
```

```python
# ==============================================================================
# 1. SINGLE INHERITANCE (One Parent -> One Child)
# ==============================================================================
class ParentA:
    def feature_a(self): return "Feature A from ParentA"

class ChildA(ParentA):
    def feature_child_a(self): return "Feature from ChildA"


# ==============================================================================
# 2. MULTIPLE INHERITANCE (Multiple Parents -> One Child)
# ==============================================================================
class Mother:
    def skills_m(self): return "Cooking skills"

class Father:
    def skills_f(self): return "Driving skills"

class ChildB(Mother, Father):
    def my_skills(self): return "Coding skills"


# ==============================================================================
# 3. MULTILEVEL INHERITANCE (Grandparent -> Parent -> Child Chain)
# ==============================================================================
class Grandparent:
    def gold(self): return "Grandparent's gold"

class ParentC(Grandparent):
    def land(self): return "Parent's land"

class ChildC(ParentC):
    def cash(self): return "Child's cash"


# ==============================================================================
# 4. HIERARCHICAL INHERITANCE (One Parent -> Multiple Independent Children)
# ==============================================================================
class Staff:
    def bio(self): return "Working at the firm"

class Manager(Staff): pass
class Clerk(Staff): pass


# ==============================================================================
# 5. HYBRID INHERITANCE (A combination of two or more types of inheritance)
# ==============================================================================
class School:
    def info(self): return "School Name"

class Student1(School): pass  # Hierarchical branch
class Student2(School): pass  # Hierarchical branch

class Alumnus(Student1, Student2): pass  # Multiple branch closing the diamond


# ==============================================================================
# CELL EXECUTION & VERIFICATION
# ==============================================================================
print("--- 1. Single Inheritance ---")
obj1 = ChildA()
print(obj1.feature_a())

print("\n--- 2. Multiple Inheritance ---")
obj2 = ChildB()
print(f"Inherited: {obj2.skills_m()} & {obj2.skills_f()}")

print("\n--- 3. Multilevel Inheritance ---")
obj3 = ChildC()
print(f"Inherited: {obj3.gold()} -> {obj3.land()}")

print("\n--- 4. Hierarchical Inheritance ---")
m, c = Manager(), Clerk()
print(f"Manager: {m.bio()} | Clerk: {c.bio()}")

print("\n--- 5. Hybrid (MRO Resolution) ---")
print(Alumnus.mro())  # Inspecting the lookup path for hybrid structure
```
