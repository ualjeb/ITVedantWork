```python
n=int(input("enter n:"))
for i in range(n,1,-1):
    for k in range(0,n-i,1):
        print(" ",end="")
    for j in range(0,i,1):
        print(j+1,end=" ")
    print()
for i in range(1,n+1,1):
    for k in range(n-i):
        print(" ",end="")
    for j in range(1,i+1,1):
        print(j,end=" ")
    print()
for i in range(n-1,0,-1):
    for k in range(0,n-i,1):
        print(" ",end="")
    for j in range(0,i,1):
        print(j+1,end=" ")
    print()
```
