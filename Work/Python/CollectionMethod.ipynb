```python
name={'Arun':15,'Ayan':16,'Anil':17}
for e in name.items():
    print(e)    #will give tuple containing key and value
for k,v in name.items():
    print(k,v)
trains = {
    "Rajdhani": 120,
    "Duronto": 95,
    "Garib Rath": 40,
    "Shatabdi": 60
}

# Dictionary comprehension
available_trains = {
    train: seats
    for train, seats in trains.items()
    if seats > 50
}

# Display result
print("Trains With More Than 50 Seats:")
print(available_trains)
```
