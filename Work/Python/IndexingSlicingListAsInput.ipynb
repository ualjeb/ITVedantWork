```python
fname=input("Enter fanme:")
print("with strip:",fname.strip())
print("split by space:",fname.split())
print("partition:",fname.partition(" "))  #default split is ' ',so for example partition will show that how it is split by showing it only one time
print("type of partition:",type(fname.partition(" ")))
```
