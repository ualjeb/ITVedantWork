```python
def ch(status_code):
    name=input('What is your name?:')
    match status_code:
        case 200:
            print("Success")
        case 404:
            print("Not Found")
        case _:  # This underscore acts as the 'default' case
            print("Unknown Error")
    return f'Status Code{status_code} Meaning found for {name}.'
sc=input('For Which status code:')
print(ch(sc))
```

# keyword Argument ----- def fun(a,b,*args,**kwargs):
here *kwargs is keyword argument, *(asterik) , variable length positional Argument, Arbitary Constant --- this will collect collection of argument
This function also called *(asterik) function.
** (double asterisk) is used for variable-length keyword arguments, A double asterisk collects an arbitrary number of named/keyword arguments (e.g. key=value) and groups them into a dictionary.

```python
def cls(*std):
    print(type(std))
    for s in std:
        print('Student Name:',s)
    else: print('Finish')
cls()
cls('Arun')
cls('Anil','Aman')
```

```python
def print_profile(**kwargs):
    print(kwargs)        # Inside the function, 'kwargs' is a dictionary
    print(type(kwargs))  # <class 'dict'>
    for i,j in kwargs.items():
        print(i,j)

# You pass named arguments
print_profile(name="Alice", age=25, role="Developer")
```

## 1. In Function Definitions (Packing)
When used inside a function's headers, * and ** "pack" multiple incoming arguments into a single data structure. By convention, programmers call them *args and **kwargs.
## 2. In Function Calls (Unpacking)
You can use the exact same symbols when calling a function to do the reverse: blow apart a collection (like a list or a dictionary) into separate, individual arguments.

```python
def total_sum(a, b, c):
    return a + b + c

my_numbers = [10, 20, 30]

# Unpacks the list items into a, b, and c automatically
result = total_sum(*my_numbers) 
print(result)
```

```python
def greet(name, age):
    print(f"Hello {name}, you are {age} years old.")

user_data = {"name": "Bob", "age": 30}

# Equates to calling: greet(name="Bob", age=30)
greet(**user_data) 
```

#### The Bare Asterisk (*)
When you place a * by itself as a parameter, it doesn't accept any values. Instead, it serves purely as a divider between positional arguments and keyword-only arguments.

```python
def create_user(username, email, *, is_admin=False, send_welcome_email=True):
    print(f"User: {username}, Admin: {is_admin}")

#  This works perfectly because keyword arguments are explicitly named
create_user("johndoe", "john@example.com", is_admin=True)

#  This throws a TypeError
#create_user("johndoe", "john@example.com", True) --- try this
# TypeError: create_user() takes 2 positional arguments but 3 were given
```

#### The *args Barrier
If your function already accepts variable positional arguments using *args, any parameter you place after *args automatically becomes a keyword-only argument. You do not need a bare asterisk in this case.

```python
def calculate_total(*prices, discount=0.0):
    total = sum(prices)
    return total * (1 - discount)

#  Python packs all numbers into *prices, and uses the keyword for discount
print(calculate_total(10, 20, 30, discount=0.1)) # Output: 54.0

#  If you try to pass the discount positionally, Python thinks it's just another price!
print(calculate_total(10, 20, 30, 0.1)) # Output: 60.1 (0.1 was added to the prices tuple)
```

## Recursion

```python
def factorial(n):
    if(n==0 or n==1):
        return 1
    else:
        return n*factorial(n-1)
factorial(5)
```

### Lambda Function
it is also known as Anonymous Function.
It is used when arguments as well as methamaical operation available only in one line.

```python
fg=lambda s,f,t : s*t+f
print(fg(4,2,8))
fg2=lambda s,f,t=1,y=2: y*t
print(fg2(1,2,3))
print(fg,fg2)
```

### Map Function
If we perform mathematical operation in list or elements available in a single list, then we use lambda and map function

```python
s=[101,102,103]
w=['A','B','C']
h=map(s,w)
print(h)
#print(list(h))
```

```python
passengers = ["Rahul", "Priya", "Amit"]
seats = ["S1", "S2", "S3"]
fares = [2500, 1800, 3200]
reservation_data = list(zip(passengers, seats, fares))
print("===================================")
print(" Railway Reservation Report ")
print("===================================")

for index, data in enumerate(reservation_data):

   passenger, seat, fare = data

   final_fare = (lambda x: x - 100)(fare)

   print("Reservation ID :", index + 1)
   print("Passenger :", passenger)
   print("Seat :", seat)
   print("Final Fare :", final_fare)
   print("-----------------------------------")
```
